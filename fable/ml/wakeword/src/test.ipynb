```python
#ran on hipergator tensorflow 2.18

import os
os.environ["CUDA_VISIBLE_DEVICES"] = "-1"
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
import tensorflow as tf
from tensorflow import keras
import tensorflow_datasets
import tensorflow_io
import librosa
from sklearn.metrics import f1_score, accuracy_score
```

```python
threshold = 0.5


def test(data="training_data_eee4773.npy", labels="training_labels_eee4773.npy", model_path="training_best_model.keras"):
    model = keras.models.load_model(model_path)

    data = np.load(data).T
    t = np.load(labels)

    X_16k = np.array([librosa.resample(x, orig_sr=48000, target_sr=16000) for x in data])

    y_pred = model.predict(X_16k)
    y_pred = (y_pred > threshold).astype(int).ravel()

    accuracy = accuracy_score(t, y_pred)
    weighted_f1 = f1_score(t, y_pred, average="weighted")

    return y_pred, accuracy, weighted_f1
```

```python
# adjust the data and labels path you want to test on; positive class means a wake word is present

y_pred, accuracy, weighted_f1 = test(
    data="./data/training_data_eee4773.npy",
    labels="./data/training_labels_eee4773.npy",
    model_path="training_best_model.keras",
)

print(f"Accuracy: {accuracy:.4f}")
print(f"Weighted F1 score: {weighted_f1:.4f}")
print(f"Labels: {y_pred}")
```
